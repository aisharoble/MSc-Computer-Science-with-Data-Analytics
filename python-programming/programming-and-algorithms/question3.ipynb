```python
# Question 3

import string


def take_input(filename="text.txt"):
    """
    (i) Read a file line-by-line using a WHILE loop.
    Return a list of lines (without trailing newline characters).
    """
    lines = []  # stores the lines from the file

    f = open(filename, "r", encoding="utf-8")  # open file
    line = f.readline()                        # read first line

    while line != "":                          # EOF when readline returns ""
        lines.append(line.rstrip("\n"))        # store without newline
        line = f.readline()                    # read next line

    f.close()  # close file
    return lines


def split_sentence(s):
    """
    (ii) Remove punctuation from a sentence, then split into words.
    """
    translator = str.maketrans("", "", string.punctuation)  # punctuation removal map
    cleaned = s.translate(translator)                       # remove punctuation
    return cleaned.split()                                  # split into words


def sort_by_length(words):
    """
    (iii) Bubble sort a list of strings by ascending length.
    Returns a NEW sorted list.
    """
    arr = words[:]  # copy
    n = len(arr)

    for _ in range(n):
        swapped = False
        for j in range(0, n - 1):
            if len(arr[j]) > len(arr[j + 1]):              # compare lengths
                arr[j], arr[j + 1] = arr[j + 1], arr[j]    # swap
                swapped = True
        if not swapped:
            break

    return arr


def process_lines(lines):
    """
    (iv) Using a WHILE loop:
    For each line: split (no punctuation), bubble sort by length, print the list.
    """
    i = 0
    while i < len(lines):
        sentence = lines[i]
        words = split_sentence(sentence)
        sorted_words = sort_by_length(words)
        print(sorted_words)  # required per line
        i += 1

# RUN/TEST
lines = take_input("text.txt")
process_lines(lines)
```
