```python
# Question 5

def input_words_until_zero():
    """
    (i) Ask for words repeatedly until the user types '0'.
    Store inputs in a set and return it (excluding '0').
    """
    inputs = set()  # will store all unique words from the user

    while True:
        w = input("Enter a word (type 0 to stop): ").strip()

        if w == "0":      # stop condition
            break

        if w != "":       # ignore empty strings
            inputs.add(w) # add word to the set

    return inputs


def words_after_threshold_case_insensitive(word_set, threshold):
    """
    (ii + iii) Validate threshold and compare case-insensitively.

    - If threshold is not a single alphabetical character:
        print 'input is incorrect' and return an empty set.

    - Otherwise:
        return a set of words starting with a letter alphabetically larger
        than threshold (case-insensitive) and print the set in the required format.
    """
    # Check that threshold is a single alphabetic character
    if not isinstance(threshold, str) or len(threshold) != 1 or not threshold.isalpha():
        print("input is incorrect")
        return set()

    t = threshold.lower()  # use lowercase for case-insensitive comparison
    output = set()

    # Loop over each word and select those starting after the threshold character
    for w in word_set:
        if w != "":
            if w[0].lower() > t:
                output.add(w)

    # Print the result in the required format
    print(f"The set of words starting from a letter larger than {threshold} is:\n")
    print(output)

    return output


def write_capitalised(word_set, filename="output.txt"):
    """
    (iv) Capitalise each word and write one per line to output.txt
    """
    with open(filename, "w", encoding="utf-8") as f:
        for w in word_set:
            f.write(w.capitalize() + "\n")  # write each word on its own line


def run_q5_program():
    """
    Full Q5 program:
      - inputs: set of words until user enters 0
      - threshold: single character input
      - output: filtered set (printed)
      - write output words to output.txt
    """
    # Get all word inputs from the user
    inputs = input_words_until_zero()

    # Ask the user for the threshold character
    threshold = input("Enter a single character threshold: ").strip()

    # Filter the set using the threshold and print the result
    output = words_after_threshold_case_insensitive(inputs, threshold)

    # Write the filtered set to output.txt using the function from Q4 part iv
    write_capitalised(output, "output.txt")


# RUN the full program
run_q5_program()
print("Wrote output words to output.txt")
```
