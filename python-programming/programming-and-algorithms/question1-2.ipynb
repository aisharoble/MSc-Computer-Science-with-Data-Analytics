```python
# Question 1

def foo(a, b):
    """
    Return a list of elements from `a` that also appear in `b`.

    Notes:
    - Preserves the order of `a`
    - Keeps duplicates from `a` (if they are also present in `b`)
    """
    l = []  # list to store matching elements

    for i in a:          # loop through each element in a
        if i in b:       # check if element is also in b
            l.append(i)  # add it to the results

    return l  # return the final list


def foo_list_comp(a, b):
    """
    Same behaviour as foo(), but written using a list comprehension.
    """
    return [i for i in a if i in b]


# RUN/TEST (optional but useful)
print(foo([1, 2, 2, 3], [2, 3]))            # expected: [2, 2, 3]
print(foo_list_comp([1, 2, 2, 3], [2, 3]))  # expected: [2, 2, 3]
```
