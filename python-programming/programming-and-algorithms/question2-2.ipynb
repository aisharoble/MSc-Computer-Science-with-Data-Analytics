```python
# Question 2

# (i) Create the dictionary and print it
capitals = {
    "United Kingdom": "London",
    "Chile": "Santiago",
    "Kazakhstan": "Astana",
    "Lebanon": "Beirut"
}

print(capitals)  # required output


# (ii) Function: return capital or print message and return None
def get_capital(country, cap_dict):
    """
    Return the capital if `country` exists in cap_dict.
    If not found: print 'Capital not found.' and return None.
    """
    if country in cap_dict:
        return cap_dict[country]

    print("Capital not found.")
    return None


# (iii) Extend: if missing, ask user, update dictionary, print confirmation
def get_capital_and_update(country, cap_dict):
    """
    Return the capital if found.
    If not found:
      - print 'Capital not found.'
      - ask user for the capital
      - update dictionary
      - print 'dictionary updated'
    """
    if country in cap_dict:
        return cap_dict[country]

    print("Capital not found.")
    new_capital = input(f"Please enter the capital for {country}: ").strip()

    if new_capital != "":
        cap_dict[country] = new_capital
        print("dictionary updated")
        return new_capital

    return None


# RUN/TEST (optional)
print(get_capital("Chile", capitals))                 # expected: Santiago
print(get_capital("France", capitals))                # prints not found, returns None
print(get_capital_and_update("France", capitals))     # asks user, updates dict
print(capitals)                                       # shows updated dictionary
```
