```python
# Question 4

def input_five_words():
    """
    (i) Ask the user for 5 words using a WHILE loop.
    Store them in a set and return the set (must end with 5 elements).
    """
    words = set()

    while len(words) < 5:           # keep asking until set contains 5 items
        w = input("Enter a word: ").strip()
        if w != "":
            words.add(w)            # add the word (set removes duplicates)

    return words


def symmetric_difference_sets(a, b):
    """
    (ii) Return the symmetric difference of two sets using set operations only.
    Implemented as: (a - b) union (b - a)
    """
    return (a - b) | (b - a)


def words_starting_after_char(word_set, ch):
    """
    (iii) Return all words that start with a character whose ASCII value
    is larger than ASCII(ch).
    """
    output = set()

    for w in word_set:
        if w != "" and ord(w[0]) > ord(ch):  # compare ASCII values
            output.add(w)

    return output


def write_capitalised(word_set, filename="output.txt"):
    """
    (iv) Capitalise each word and write them one per line to output.txt
    """
    with open(filename, "w", encoding="utf-8") as f:
        for w in word_set:
            f.write(w.capitalize() + "\n")   # write each capitalised word


# -------------------------
# RUN / TEST SECTION
# -------------------------

# Ask user for the first set of 5 words
set1 = input_five_words()

# Ask user for the second set of 5 words
set2 = input_five_words()

# Compute the symmetric difference between the two sets
sym_diff = symmetric_difference_sets(set1, set2)
print("Symmetric difference:", sym_diff)

# Ask the user for a character threshold
ch = input("Enter a character: ").strip()

# Filter words whose first letter is alphabetically larger than ch
filtered = words_starting_after_char(sym_diff, ch)
print("Filtered set:", filtered)

# Write the filtered, capitalised words to output.txt
write_capitalised(filtered, "output.txt")
print("Wrote filtered words to output.txt")
```
