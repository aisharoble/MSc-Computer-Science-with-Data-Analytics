```python
class Weight_graph:
    """
    Represent a directed weighted graph using an adjacency list.
    Each vertex maps to a list of (neighbour, weight) pairs.
    """

    def __init__(self):
        # Dictionary: vertex -> list of (neighbour, weight)
        self.adj_list = {}

    def add_vertex(self, v):
        """
        Add a vertex to the graph if it does not already exist.
        """
        if v not in self.adj_list:
            self.adj_list[v] = []

    def add_edge(self, from_v, to_v, weight):
        """
        Add a directed, weighted edge from from_v to to_v.
        Vertices are created automatically if needed.
        """
        self.add_vertex(from_v)
        self.add_vertex(to_v)
        self.adj_list[from_v].append((to_v, weight))

    def __str__(self):
        """
        Return a string representation of the graph's edges.
        """
        lines = []
        for v, neighbours in self.adj_list.items():
            for (n, w) in neighbours:
                lines.append(f"{v} --{w}--> {n}")
        return "\n".join(lines)

    def __invert__(self):
        """
        Reverse the direction of all edges in the graph.

        Implements the bitwise not operator (~) for Weight_graph.
        Returns a new Weight_graph object containing all vertices
        and edges reversed: for every edge u --w--> v, we create
        v --w--> u.
        """
        # Create a new empty graph
        reversed_graph = Weight_graph()

        # First add all vertices to the new graph
        for v in self.adj_list:
            reversed_graph.add_vertex(v)

        # Now reverse every edge
        for from_v, neighbours in self.adj_list.items():
            for (to_v, weight) in neighbours:
                # Add reversed edge to_v -> from_v with same weight
                reversed_graph.add_edge(to_v, from_v, weight)

        return reversed_graph
```

```python
# Create the weighted graph from the diagram
g = Weight_graph()

# Add the directed, weighted edges
g.add_edge("J", "G", 5)
g.add_edge("J", "L", 7)
g.add_edge("G", "B", 2)
g.add_edge("G", "N", 4)
g.add_edge("N", "G", 4)
g.add_edge("B", "C", 3)
g.add_edge("C", "L", 6)
g.add_edge("B", "L", 8)

print("Original graph edges:")
print(g)
print("Original graph:")
print(g)

# Use the overloaded ~ operator to reverse all edge directions
reversed_g = ~g

print("\nReversed graph:")
print(reversed_g)
```

### Q5 (ii) – Breadth-first traversal from G

We perform BFS using a queue and a visited set.
We only follow edges in their arrow direction.

Initial state:
- Queue: [G]
- Visited: {G}
- Visiting order so far: [ ]

Step 1:
- Dequeue G → visit G
- Neighbours of G: B, N
- Add unvisited neighbours to the queue: enqueue B, N
- Queue: [B, N]
- Visited: {G, B, N}
- Visiting order: [G]

Step 2:
- Dequeue B → visit B
- Neighbours of B: C
- C is not visited, so enqueue C
- Queue: [N, C]
- Visited: {G, B, N, C}
- Visiting order: [G, B]

Step 3:
- Dequeue N → visit N
- Neighbours of N: G
- G is already visited, so we do not enqueue it again
- Queue: [C]
- Visited: {G, B, N, C}
- Visiting order: [G, B, N]

Step 4:
- Dequeue C → visit C
- Neighbours of C: L
- L is not visited, so enqueue L
- Queue: [L]
- Visited: {G, B, N, C, L}
- Visiting order: [G, B, N, C]

Step 5:
- Dequeue L → visit L
- Neighbours of L: B
- B is already visited, so no new vertices are added
- Queue: [ ]
- Visited: {G, B, N, C, L}
- Visiting order: [G, B, N, C, L]

The BFS traversal starting from G visits the vertices in the order:
**[G, B, N, C, L]**

Vertex J is not reached because all edges involving J point *towards* G or L,
and there is no edge from G back to J.
