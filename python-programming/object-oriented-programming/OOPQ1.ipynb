```python
class Weather:
    """
    Represent the weather for a given temperature and
    provide a method to describe the day as hot, cold or mild.
    """

    def __init__(self, temperature):
        """
        Initialise a Weather object with a temperature value.
        """
        # Store the temperature in an instance variable
        self.temperature = temperature

    def check_weather(self):
        """
        Print a message describing the kind of day based on the temperature.
        """
        # Use the stored temperature to decide what kind of day it is
        if self.temperature >= 30:
            print("Hot Day")
        elif self.temperature >= 15:
            print("Mild Day")
        else:
            print("Cold Day")
# Given starting temperature from the question
current_temp = 18

# Create a Weather object using the current temperature
today_weather = Weather(current_temp)

# Call the method to display the type of day
today_weather.check_weather()
```

This class-based version stores the temperature as an instance variable. The check_weather method reproduces the behaviour of the original function by printing whether the day is hot, mild or cold.
