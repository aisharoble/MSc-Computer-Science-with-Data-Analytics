### Q3 (i) – Heap sort by hand

Initial list:
["4. Delta", "2. Beta", "3. Gamma", "1. Alpha", "5. Epsilon"]

#### Building the max-heap
1. Compare "2. Beta" ↔ "5. Epsilon" → swap because 5 > 2  
   → ["4. Delta", "5. Epsilon", "3. Gamma", "1. Alpha", "2. Beta"]  
2. Compare "4. Delta" ↔ "5. Epsilon" → swap because 5 > 4  
   → ["5. Epsilon", "4. Delta", "3. Gamma", "1. Alpha", "2. Beta"]

Heap built: **["5. Epsilon", "4. Delta", "3. Gamma", "1. Alpha", "2. Beta"]**

#### Extraction phase
1. Extract 5 (Epsilon), swap with last → ["2. Beta", "4. Delta", "3. Gamma", "1. Alpha", "5. Epsilon"]  
   Re-heapify → ["4. Delta", "2. Beta", "3. Gamma", "1. Alpha", "5. Epsilon"]  
   Sorted part: ["5. Epsilon"]

2. Extract 4 (Delta) → ["1. Alpha", "2. Beta", "3. Gamma", "4. Delta", "5. Epsilon"]  
   Re-heapify → ["3. Gamma", "2. Beta", "1. Alpha", "4. Delta", "5. Epsilon"]  
   Sorted part: ["4. Delta", "5. Epsilon"]

3. Continue until sorted → Final ascending list:  
   **["1. Alpha", "2. Beta", "3. Gamma", "4. Delta", "5. Epsilon"]**


```python
def extract_key(s):
    """
    Extract the integer at the start of a string like '4. Delta'.
    """
    return int(s.split(".")[0])

def heapify(data, n, i):
    """
    Maintain the max-heap property for the subtree rooted at index i.

    :param data: list of strings like '4. Delta'
    :param n: size of the heap (may be smaller than len(data) during sort)
    :param i: index of the current root of the subtree
    """
    largest = i          # Assume current index i is the largest
    left = 2 * i + 1     # Left child index
    right = 2 * i + 2    # Right child index

    # Check if left child exists and is greater than current largest
    if left < n and extract_key(data[left]) > extract_key(data[largest]):
        largest = left

    # Check if right child exists and is greater than current largest
    if right < n and extract_key(data[right]) > extract_key(data[largest]):
        largest = right

    # If largest has changed, swap and continue heapifying
    if largest != i:
        # Swap the root with the larger child
        data[i], data[largest] = data[largest], data[i]
        # Recursively heapify the affected subtree
        heapify(data, n, largest)
class Heap:
    """
    Simple max-heap wrapper for strings like '4. Delta'.
    The underlying list 'data' is kept as a valid heap using heapify
    and the extract_key function for comparisons.
    """

    def __init__(self):
        # Underlying Python list that stores the heap elements
        self.data = []

    def insert(self, value):
        """
        Insert a new value into the heap and restore the heap property.
        """
        # Add the new value to the end of the list
        self.data.append(value)

        # Rebuild the heap so that 'data' remains a valid max-heap
        n = len(self.data)
        for i in range(n // 2 - 1, -1, -1):
            heapify(self.data, n, i)


def heap_sort(data):
    """
    Perform heap sort on a list of strings in-place.

    The list elements are strings in the format '<number>. <text>'.
    The list is sorted in ascending order based on the numeric prefix.
    """
    n = len(data)

    # 1) Build a max-heap from the list
    for i in range(n // 2 - 1, -1, -1):
        heapify(data, n, i)

    # 2) Extract elements from the heap one by one
    for i in range(n - 1, 0, -1):
        # Move current root (largest) to the end of the list
        data[0], data[i] = data[i], data[0]

        # Call heapify on the reduced heap
        heapify(data, i, 0)
```

```python
class Greek_numbers:
    """
    Represent a collection of Greek-number strings stored in a heap.
    """

    def __init__(self, values):
        """
        Construct a Greek_numbers object from a list of strings.
        :param values: list of strings like '4. Delta'
        """
        # Create an empty heap instance (Heap class from Appendix A)
        self.heap = Heap()

        # Insert each value into the heap to maintain heap order
        for v in values:
            self.heap.insert(v)
```

```python
greek_list = ["4. Delta", "2. Beta", "3. Gamma", "1. Alpha", "5. Epsilon"]

# Part (iii): create Greek_numbers object
greek_numbers = Greek_numbers(greek_list)

# Part (ii): demonstrate heap_sort
print("Before sort:", greek_list)
heap_sort(greek_list)
print("After sort: ", greek_list)

```
