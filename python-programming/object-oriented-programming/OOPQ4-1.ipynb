```python
class Node:
    """
    Represent a single node in a binary tree.
    Each node stores a value and references to its left and right children.
    """
    def __init__(self, value):
        # Store the value for this node
        self.value = value
        # Left and right child pointers, start as None
        self.left = None
        self.right = None
```

```python
# Build the tree shown in the diagram

root = Node(8)          # top node

# Left subtree of 8
root.left = Node(5)
root.left.left = Node(4)
root.left.right = Node(6)

# Right subtree of 8
root.right = Node(12)
root.right.left = Node(5)
root.right.right = Node(20)
root.right.right.left = Node(17)

# This tree is NOT a binary search tree.
# Justification:
# - In a BST, every value in the right subtree of the root (8) should be > 8.
# - However, there is a node with value 5 in the right subtree of 8 (left child of 12),
#   and 5 < 8, so this breaks the binary search tree property.
```

### Q4 (ii) – Postorder traversal

Postorder visits nodes in the order: left subtree, right subtree, then the node itself.

- Left subtree of 8 (rooted at 5):
  - Visit 4 (left child)
  - Visit 6 (right child)
  - Visit 5 (their parent)
  → [4, 6, 5]

- Right subtree of 8 (rooted at 12):
  - Visit 5 (left child of 12)
  - Visit 17 (left child of 20)
  - Visit 20 (parent of 17)
  - Visit 12
  → [5, 17, 20, 12]

Finally, visit the root 8.

Overall postorder visiting order:
[4, 6, 5, 5, 17, 20, 12, 8]


```python
class Search_tree:
    """
    Binary search tree where the root always has value 0.
    Odd numbers are inserted into the left subtree of the root.
    Even numbers are inserted into the right subtree of the root.
    Each subtree then follows the normal binary search tree rules.
    """

    def __init__(self, values=None):
        """
        Construct a Search_tree.
        If a list of values is provided, it is used to build the tree.
        Non-integer values are discarded with a message.
        """
        # Root node always starts with value 0
        self.root = Node(0)

        # If values are provided, insert them one by one
        if values is not None:
            for v in values:
                # Check that the value is an integer
                if not isinstance(v, int):
                    # Display a message and skip non-integer values
                    print(f"Value {v} is not an integer and will be discarded.")
                    continue

                # Insert valid integer values using the special rules
                self.insert(v)

    def insert(self, value: int):
        """
        Insert an integer into the tree.
        Odd values go into the left subtree of the root.
        Even values (other than 0) go into the right subtree of the root.
        """
        # Do not insert another 0, root already stores 0
        if value == 0:
            return

        # Choose subtree based on odd or even
        if value % 2 == 1:
            # Insert into the left subtree of the root
            self.root.left = self._insert_bst(self.root.left, value)
        else:
            # Insert into the right subtree of the root
            self.root.right = self._insert_bst(self.root.right, value)

    def _insert_bst(self, node, value: int):
        """
        Helper method to insert a value into a normal binary search tree.

        :param node: current root node of the subtree
        :param value: integer value to insert
        :return: root node of the updated subtree
        """
        # If there is no node here yet, create one
        if node is None:
            return Node(value)

        # Standard BST rules within the chosen subtree
        if value < node.value:
            node.left = self._insert_bst(node.left, value)
        else:
            node.right = self._insert_bst(node.right, value)

        return node
```
